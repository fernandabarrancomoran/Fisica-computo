```python
import math as mt
```

```python
r = float(input('introduce r'))
theta = float(input('introduce el angulo theta'))
theta = mt.radians(theta)
x =r * mt.cos(theta)
y =r * mt.sin(theta) 
print("x=",x)
print("y=",y)
```

a) Escribe un programa que tome como entrada los valores de A y Z, e imprima la
energ ́ıa de enlace B para el atomo correspondiente. Usa tu programa para encontrar  ́
la energ ́ıa de enlace de un atomo con  ́ A = 58 y Z = 28 (Hint: La respuesta correcta
es alrededor de los 490 MeV).

```python
import numpy as np

a1=15.8
a2=18.3
a3=0.714
a4=23.2
A = float(input('introduce el numero atomico'))
Z = float(input('introduce el numero de protones'))
if ( A%2!=0 ):
    a5=0
elif (A%2==0 and Z%2==0):
    a5=12
elif (A%2!=0 and Z%2!=0):
    a5=-12
B=(a1*A)-(a2*(A**(2/3)))-(a3*((Z**2)/(A**(1/3))))-(a4*(((A-(2*Z))**2)/A))+a5/(A**(1/2))
print("la energia de enlace nuclear es", B , "MeV")
```
